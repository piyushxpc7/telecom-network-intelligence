# Telecom EDA Notebook

Phase 1 exploratory analysis for mobile activity data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = "../data/raw/sample.csv"  # update this path
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
df.dtypes, df.describe(include='all'), df.isnull().sum()
```

```python
df.columns = [c.strip().lower() for c in df.columns]
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')

missing_timestamps = df['timestamp'].isna().sum()
negative_values = {
    'call_count': int((pd.to_numeric(df['call_count'], errors='coerce') < 0).sum()),
    'sms_count': int((pd.to_numeric(df['sms_count'], errors='coerce') < 0).sum()),
    'internet_usage': int((pd.to_numeric(df['internet_usage'], errors='coerce') < 0).sum()),
}
duplicates = int(df.duplicated().sum())

print('missing_timestamps:', missing_timestamps)
print('negative_values:', negative_values)
print('duplicate_rows:', duplicates)
```

```python
df['day'] = df['timestamp'].dt.date
daily = df.groupby('day', as_index=False)[['call_count', 'sms_count', 'internet_usage']].sum()

plt.figure(figsize=(12, 5))
plt.plot(daily['day'], daily['call_count'], label='Calls')
plt.plot(daily['day'], daily['sms_count'], label='SMS')
plt.plot(daily['day'], daily['internet_usage'], label='Internet')
plt.xticks(rotation=45)
plt.legend()
plt.title('Usage over time')
plt.tight_layout()
plt.show()
```

```python
region_dist = df.groupby('grid_id', as_index=False)['internet_usage'].sum().sort_values('internet_usage', ascending=False).head(20)
plt.figure(figsize=(12, 5))
sns.barplot(data=region_dist, x='grid_id', y='internet_usage')
plt.xticks(rotation=60)
plt.title('Internet usage distribution by region (top 20)')
plt.tight_layout()
plt.show()
```
